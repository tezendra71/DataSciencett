```python
import pandas as pd
import numpy as np
import random

# Generating customer IDs
customer_ids = range(1, 501)

# Generating genders
genders = np.random.choice(['Male', 'Female', 'Non-Binary'], size=500, p=[0.45, 0.45, 0.10])

# Sample cities, states, and countries
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio', 'San Diego', 'Dallas', 'San Jose']
states = ['New York', 'California', 'Illinois', 'Texas', 'Arizona', 'Pennsylvania', 'Texas', 'California', 'Texas', 'California']
countries = ['USA'] * 500  # All customers from USA

# Randomly assign cities and corresponding states
random.seed(0)
city_state_pairs = random.choices(list(zip(cities, states)), k=500)

# Split city_state_pairs into separate city and state lists
customer_cities = [pair[0] for pair in city_state_pairs]
customer_states = [pair[1] for pair in city_state_pairs]

# Create the DataFrame
customer_demographics = pd.DataFrame({
    'customer_id': customer_ids,
    'gender': genders,
    'city': customer_cities,
    'state': customer_states,
    'country': countries
})

# Display the first few rows of the dataset
print(customer_demographics.head())

# Save to CSV
customer_demographics.to_csv('customer_demographics.csv', index=False)
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Number of transactions to generate
num_transactions = 1000

# Generating transaction IDs
transaction_ids = range(1, num_transactions + 1)

# Generating customer IDs as Foreign Keys
customer_ids = np.random.choice(range(1, 501), size=num_transactions)

# Generating random transaction dates within the last year
start_date = datetime.now() - timedelta(days=365)
transaction_dates = [start_date + timedelta(days=random.randint(0, 365)) for _ in range(num_transactions)]

# Generating random transaction amounts between $5 and $500
transaction_amounts = np.random.uniform(5, 500, size=num_transactions).round(2)

# Generating random payment methods
payment_methods = np.random.choice(['Credit Card', 'Debit Card', 'Cash', 'PayPal'], size=num_transactions)

# Generating random product IDs (assuming we have 50 products)
product_ids = np.random.choice(range(1, 51), size=num_transactions)

# Generating random quantities between 1 and 10
quantities = np.random.randint(1, 11, size=num_transactions)

# Create the DataFrame
transaction_history = pd.DataFrame({
    'transaction_id': transaction_ids,
    'customer_id': customer_ids,
    'transaction_date': transaction_dates,
    'transaction_amount': transaction_amounts,
    'payment_method': payment_methods,
    'product_id': product_ids,
    'quantity': quantities
})

# Display the first few rows of the dataset
print(transaction_history.head())

# Save to CSV
transaction_history.to_csv('transaction_history.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import random

# Number of products to generate
num_products = 50

# Generating product IDs
product_ids = range(1, num_products + 1)

# Sample categories
categories = ['Electronics', 'Clothing', 'Home & Kitchen', 'Sports', 'Beauty', 'Toys']
product_categories = np.random.choice(categories, size=num_products)

# Generating random prices between $10 and $1000
prices = np.random.uniform(10, 1000, size=num_products).round(2)

# Generating random stock quantities between 0 and 100
stock_quantities = np.random.randint(0, 101, size=num_products)

# Create the DataFrame
product_catalog = pd.DataFrame({
    'product_id': product_ids,
    'category': product_categories,
    'price': prices,
    'stock_quantity': stock_quantities
})

# Display the first few rows of the dataset
print(product_catalog.head())

# Save to CSV
product_catalog.to_csv('product_catalog.csv', index=False)
```
